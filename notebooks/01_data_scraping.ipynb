# Tahap 1: Scraping Ulasan Google Play Store (Tokopedia)

Notebook ini mengambil ulasan aplikasi **Tokopedia** (`com.tokopedia.tkpd`) dari Google Play Store berbahasa Indonesia menggunakan `google-play-scraper`, membersihkan duplikat/nilai kosong, lalu menyimpannya sebagai `ulasan_playstore.csv`.

```python
import pandas as pd
from google_play_scraper import Sort, reviews

# Konfigurasi parameter scraping
APP_ID = 'com.tokopedia.tkpd'
TARGET_COUNT = 15000

print(f"Memulai proses scraping {TARGET_COUNT} ulasan untuk {APP_ID}...")

result, continuation_token = reviews(
    APP_ID,
    lang='id',
    country='id',
    sort=Sort.NEWEST,
    count=TARGET_COUNT,
    filter_score_with=None,
)

df_reviews = pd.DataFrame(result)
columns_to_keep = ['reviewId', 'userName', 'content', 'score', 'thumbsUpCount', 'at']
df_clean = df_reviews[columns_to_keep].copy()

df_clean.dropna(subset=['content'], inplace=True)
df_clean.drop_duplicates(subset=['content'], inplace=True)

print(f"Scraping selesai! Jumlah data bersih yang didapat: {len(df_clean)} sampel.")

OUTPUT_FILENAME = 'ulasan_playstore.csv'
df_clean.to_csv(OUTPUT_FILENAME, index=False, encoding='utf-8')
print(f"Dataset berhasil disimpan ke dalam file: {OUTPUT_FILENAME}")

df_clean.head()
```
